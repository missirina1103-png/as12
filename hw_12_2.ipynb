```python
import numpy as np

# 2.1
np.random.seed(33)
inc = np.random.uniform(100, 1000, size=(3, 10))
inc = np.round(inc, 2)

print("2.1","\n", "Without tax","\n", np.round(inc, 2))

# 2.2
inc_30 = inc*0.7

print("2.2","\n", "30% tax","\n", np.round(inc_30, 2))

# 2.3
inc_new = inc_30.copy()
inc_new[0:3, 4:7]=inc_new[0:3, 4:7]/0.7*0.6

print("2.3","\n", "New tax system", "\n", np.round(inc_new, 2))

# 2.4
print("2.4","\n", "2 quarter", "\n", np.round(inc_new[:,3:6], 2))
```
