```python
import numpy as np

np.random.seed(33)
np.set_printoptions(precision=2)

# 2.1
inc = np.random.uniform(100, 1000, size=(3, 10)).round(2)
print(f"2.1 Without tax\n{inc}\n")

# 2.2
print(f"2.2 30% tax\n{inc * 0.7}\n")

# 2.3 tax is 40% in May-Jul and 30% in the other months
tax = np.full(10, 0.3)
tax[4:7] = 0.4
inc_new = inc * (1 - tax)
print(f"2.3 New tax system\n{inc_new}\n")

# 2.4
print(f"2.4 2 quarter\n{inc_new[:, 3:6]}")
```
